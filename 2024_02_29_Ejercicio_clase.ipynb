```python
import numpy as np
from random import random as rd

def func():
  r = rd()
  pi = 0.5
  s = pi

  i = 1
  while s < r:
    pi = pi/2
    s += pi
    i += 1
  #print(k)
  return i

def f(x):
  if 0 <= x < 0.1:
    return 1
  elif 0.1 <= x < 0.3:
    return 2
  elif 0.3 <= x < 0.7:
    return 3
  elif 0.7 <= x <= 1:
    return 4
  else:
    return 0

def h(x, a):
  if x > a:
    return 1
  else:
    return 0


if __name__ == "__main__":
  #x = bino(10, 1/2)
  #print(x)
  #print(2+2)
  n = 1000000
  l = list()
  for i in range(n):
    r = rd()
    l.append(f(r))
  l = np.array(l)

  #print(l)
  #print(l.mean())
  #print(l.var() / n, "\n")

  #a)---------------------------------------------------------------------------

  g = lambda x: 1/(1+x)
  e = []
  for i in range(n):
    r = rd()
    e.append(g(f(r)))

  print("Ejercicio 1:\nInciso a): \nE[1/(1+x)] = ", np.mean(e))
  print("V[1/(1+x)] = ", np.var(e)/n, "\n")

  #b)---------------------------------------------------------------------------
  p = 0
  for i in range(n):
    r = rd()
    p += h(f(r)**2 + f(r),5)
  print("Inciso b) \nP(x^2 + x > 5): ", p/n, "\n")

  #2.---------------------------------------------------------------------------
  x = list()
  for i in range(n):
    x.append(np.sqrt(func()))
  print("\nEjercicio 2: \nE[sqrt(x)] = ", np.mean(x))
  print("V[sqrt(x)] = ",np.var(x)/n)
```
