```python
from random import random, choices, gauss
import matplotlib.pyplot as plt
import numpy as np
from numba import jit

#@jit(nopython = True)
def hola(n=10**3, m=10**4):
  l = []
  for i in range(n):
    #l.append(-np.log(random())/2)
    #l.append(gauss(5, 2))
    l.append(random())

  ls = []
  for i in range(m):
    l1 = choices(l, k = m)
    ls.append(np.var(l1))
  return ls

ls = hola(1000, 10000)
li = np.percentile(ls, 5)
ld = np.percentile(ls, 95)

print(li, ld)
print(np.mean(ls))
plt.hist(ls, density = True, bins = 20, color = 'r', edgecolor = 'k')
plt.show()
```
