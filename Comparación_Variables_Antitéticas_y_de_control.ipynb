#Variables antitéticas

```python
from random import random
import numpy as np
from numba import jit
from scipy.stats import pearsonr
```

```python
@jit(nopython = True)
def f(nr = 10**6):
  l = []
  for i in range(nr):
    s, i = 0, 0
    while s < 1:
      s += random()
      i += 1
    l.append(i)
  return l
```

```python
@jit(nopython = True)
def g(nr = 10**6):
  l = []
  for i in range(nr):
    s1, s2, i = 0, 0, 0
    while s1 < 1 or s2 < 1:
      r = random()
      if s1 < 1:
        s1 += r
        i += 1
      if s2 < 1:
        s2 += 1-r
        i += 1
    l.append(i/2)
  return l
```

```python
nr = 10**6
l1 = f(nr)
l2 = g(nr)
m1 = np.mean(l1)
v1 = np.var(l1, ddof = 1)

m2 = np.mean(l2)
v2 = np.var(l2, ddof = 1)

print(m1, v1)
print(m2, v2)

print(str(round(100 * (v1 - v2)/v1, 3))+'%')
```

#Variables de control

```python
@jit(nopython = True)
def h(nr = 10**6):
  li, ls = [], []
  for i in range(nr):
    s, i, s1 = 0, 0, 0
    while s < 1:
      r = random()
      s += r
      s1 += np.e / i
      i += 1
    li.append(i)
    ls.append(r)
  return li, ls
```

```python
nr = 10**6
a, b = h(nr)
print(100 * pearsonr(a,b)[0]**2)
```
