```python
import numpy as np
from random import random
import matplotlib.pyplot as plt
from numba import jit
```

```python
def f(x):
  return np.sqrt(1 + x**3)
```

```python
def g(x):
  # return 1/(1+x) # np.log(2)
  return np.sqrt(x**3) # 5/7
```

```python
@jit(nopython = True)
def anti(n):
  la, lb, lc = [], [], []
  for i in range(n):
    r = random()
    a = np.sqrt(1 + r**3)
    b = np.sqrt(1 + (1-r)**3)
    c = (a + b)/2
    la.append(a)
    lb.append(b)
    lc.append(c)
  return la, lb, lc
```

```python
@jit(nopython = True)
def control(n, e):
  la, lb = [], []
  for i in range(n):
    r = random()
    a = np.sqrt(1 + r**3)
    b = np.sqrt(r**3)
    la.append(a)
    lb.append(b)
  co = np.cov(la,lb)
  cv = -co[0,1]/co[1,1]

  la, lb, lc = [], [], []
  for i in range(n):
    r = random()
    a = np.sqrt(1 + r**3)
    b = np.sqrt(r**3)
    c = a + cv*(b - e)
    la.append(a)
    lb.append(b)
    lc.append(c)
  return la, lb, lc, co
```

```python
a, b, c = anti(10**6)
ma, mb, mc = np.mean(a), np.mean(b), np.mean(c)
va, vb, vc = np.var(a, ddof = 1), np.var(b, ddof = 1), np.var(c, ddof = 1)

print(ma, mb, mc)
print(va, vb, vc,'\n')

red = 100*(va-vc)/va
print(f'La reducciòn de varianza porcentual es: {round(red, 5)}%')
```

```python
a, b, c, co = control(10**6, 2/5)
print(co)
ma, mb, mc = np.mean(a), np.mean(b), np.mean(c)
va, vb, vc = np.var(a, ddof = 1), np.var(b, ddof = 1), np.var(c, ddof = 1)

print(ma, mb, mc)
print(va, vb, vc,'\n')

red = 100*(va-vc)/va
print(f'La reducciòn de varianza porcentual es: {round(red, 5)}%')
```

```python
@jit(nopython = True)
def control1(n, e):
  la, lb = [], []
  for i in range(n):
    r = random()
    a = np.sqrt(1 + r**3)
    b = np.sqrt(r**5)
    la.append(a)
    lb.append(b)
  co = np.cov(la,lb)
  cv = -co[0,1]/co[1,1]

  la, lb, lc = [], [], []
  for i in range(n):
    r = random()
    a = np.sqrt(1 + r**3)
    b = np.sqrt(r**5)
    c = a + cv*(b - e)
    la.append(a)
    lb.append(b)
    lc.append(c)
  return la, lb, lc, co
```
