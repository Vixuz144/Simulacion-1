```python
import numpy as np
from random import random
from numba import jit
```

```python
@jit(nopython = True)
def f(n = 10**4):
  l=[]
  for i in range(n):
    r = random()
    # print(r)
    pk = 1/2
    suma = pk
    k = 1
    while r > suma:
      pk = k/(k+2) * pk
      suma = suma + pk
      k=k+1
    # print(k)
    l.append(k)
  l = np.array(l)
  return l
```

```python
print(np.mean(1/f()**2))
```

```python
print(f(100000))
```

```python
def ej19(lam):
  r = random()
  return -np.log(1-r)/lam

l = []
for i in range(100000):
  v = ej19(0.5)
  l.append(v)
np.mean(l)
```

```python
from random import random
from statistics import mean, variance

def bino(u):
    pb = 1/2
    suma = pb
    k = 1
    while u > suma:
        pb = k/(k+2)*pb
        suma = suma + pb
        k = k + 1
    return k

nr = 100000

bi = []
for i in range(nr):
    u = random()
    x = bino(u)
    bi.append(1/x**2)

print(mean(bi))
```
