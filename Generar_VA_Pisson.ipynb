```python
from random import random as rd
from math import exp
from statistics import mean, variance
from numba import jit

@jit(nopython = True)
def poisson(l):
  r = rd()
  pk = exp(-l)
  s = pk
  k=0
  while s<r:
    pk = l*pk/(k+1)
    s += pk
    k += 1
  return k

lamb= 10
n = 10000000
l = list()
for i in range(n):
  l.append(poisson(lamb))
print(mean(l), variance(l)/n)
```
