```python
import numpy as np
from random import random as rd
from numba import jit

#@jit(nopython = True)
def int_MC(f,n):
  s = 0
  for i in range(n):
    x=rd()
    s+=f(x)
  return s/n

#@jit(nopython = True)
def int_MC_list(f,n):
  s = []
  for i in range(n):
    x=rd()
    s.append(f(x))
  return s

if __name__== "__main__":
  f = lambda x: 1/(1+x)
  #print(f(2))
  n=1000000
  u = int_MC(f,n)
  v = np.log(2)
  print(u,v,u-v)

  w = np.array(int_MC_list(f,n))
  print(np.mean(w), np.var(w), np.var(w)/n)

  g = lambda x: 5/(3+5*x)
  n=1000000

  a = int_MC_list(g,n)
  print(np.mean(a), np.var(a), np.var(a)/n)

  #3

  h = lambda z: (np.e**(-(1/z-1)**2))/z**2

  b = int_MC_list(h,n)
  print(np.mean(b), np.var(b), np.var(b)/n)
```
