```python
from random import expovariate
import matplotlib.pyplot as plt
import numpy as np

def f():
  la = 3
  lb = 5

  ta = 0
  tb = 0.2

  tr = 30
  periodicidad = 30

  cap = 0
  t=0

  lx = [t]
  ly = [cap]

  while min(ta,tb,tr) < 365:
    t=min(ta,tb,tr)
    lx.append(t)
    ly.append(cap)
    if t==ta:
      cap += 80
      #print("venta del producto A",t,cap)
      ta = t + expovariate(la)
    elif t == tb:
      cap += 100
      #print("venta del producto B",t,cap)
      tb = t + expovariate(lb)
    else: # t == tr
      cap -= 3000
      #print("pago de renta",t,cap)
      ta = t + expovariate(la)
      tr = t + periodicidad
  #print(cap)
  return lx, ly, cap

n = 10000
cap = []
for i in range(n):
  a, b, c = f()
  cap.append(c)

print(np.mean(cap))
plt.hist(cap, density = True, color = "b", edgecolor = "k")
plt.show()
```

```python
n = 1000
cap = []
cont = 0
for i in range(n):
  a, b, c = f()
  cap.append(c)
  if c < 230000:
    cont += 1

print(np.mean(cap), cont/n)
plt.hist(cap, density = True, color = "b", edgecolor = "k")
plt.show()
```
