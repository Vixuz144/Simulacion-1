```python
from scipy.stats import expon
import numpy as np
from random import random
```

```python
def linea_espera(T, ll, ls):
  n = len(ls)
  t = 0
  l = 0

  tl = -np.log(random())/ll
  ts = np.inf * np.ones(n)

  lin = np.linspace(0,1,n+1)

  while t < T:
    t = min(tl, np.min(ts))

    if t == tl: #Llegada
      l += 1
      tl += -np.log(random())/ll

      if l <= n: #n servidores, al menos uno vacio
        for i in range(n):
          a = lin[i]
          b = lin[i+1]
          if a < random() <=b and np.isinf(ts[i]):
            des = i
            ts[des] = t - np.log(random())/ls[des]
      else:
        des = int(list(np.where(ts == min(ts))[0])[0])
        ts[des] = t - np.log(random())/ls[des]
      print("%10.3f"%t, l, des+1)

    else: #Salida
      l -= 1

      des = int(list(np.where(ts == min(ts))[0])[0])
      if l > 1:
        ts[des] = t + expon.rvs(ls[des])
      else:
        ts[des] = np.inf
      print("\t\t","%10.3f"%t, l, des+1)
```
