```python
from random import gauss
import numpy as np
import matplotlib.pyplot as plt
```

```python
def trayectoria(p0, mu, sigma, T):
  l = [p0]
  p = p0
  for i in range(T):
    p= p*np.exp(mu + sigma*gauss(0,1))
    l.append(p)
  return l
```

```python
o = 0.002
s = 0.01
T = 30
p = 10
for i in range(10):
  l=trayectoria(p,o,s,T)
  plt.plot(l)
plt.show()
```

```python
def precio(p, mu, sigma,T):
  return p*np.exp(mu*T + sigma*np.sqrt(T)*gauss(0,1))
```

```python
rep = 100000
L=[]
for i in range(rep):
  L.append(precio(p,o,s,T))

plt.hist(L, bins = 20, density = True, edgecolor = "k", color = 'w')
plt.show()
```

```python
t1 = 5
t2 = 10
p= 10
c = 0
for i in range(rep):
  p5 = precio(p,o,s,5)
  p10 = precio(p5, o, s,5)
  c+=100*(p10-p5)/p5
print(c/rep)
```
