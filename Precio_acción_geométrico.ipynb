```python
from random import gauss, random
import numpy as np
import matplotlib.pyplot as plt
from numba import jit
```

```python
@jit(nopython = True)
def trayectoria(p0, mu, sigma, T):
  l = [p0]
  p = p0
  for i in range(T):
    p = p*np.exp(mu + sigma*gauss(0,1))
    l.append(p)
  return l
```

```python
mu = 0.001
s = 0.03
p0 = 10
t = 30
l = []
rep = 10**6
for i in range(rep):
  l.append(trayectoria(p0, mu, s, t)[-1])

print("valor simulado:",np.mean(l))
print("valorexacto:", p0*np.exp(mu*t + (s**2)*t/2))
```

```python
def precio(p, mu, sigma,T):
  return p*np.exp(mu*T + sigma*np.sqrt(T)*gauss(0,1))
```

```python
t1 = 50
t2 = 10
rep = 10**6
l = []
for i in range(rep):
  p10 = precio(p0,mu,s,t2)
  p50 = precio(p10,mu,s,t1-t2)
  l.append(100 * (p50-p10)/p10)

print(np.mean(l))
```

```python
@jit(nopython = True)
def prob(p0, mu, s, t, pr, rep):
  c = 0
  for i in range(rep):
    p = p0*np.exp(mu*t + s*np.sqrt(t)*gauss(0,1))
    if p>pr:
      c+=1
  return c/rep
```

```python
pr = 10.5

print(prob(p0,mu,s,t,pr,rep))
```
