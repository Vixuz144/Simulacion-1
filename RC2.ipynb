```python
from random import random
import numpy as np
from numba import jit
```

```python
@jit(nopython = True)
def rc2(n = 10**6):
  l = []
  for i in range(n):
    r = (2*random())**2
    if r < 2:
      l.append(2)
    else:
      l.append(0)
  return l
```

```python
n = 10**7
l = rc2(n)
print(np.mean(l))
print(np.var(l,ddof = 1)/n)
```
