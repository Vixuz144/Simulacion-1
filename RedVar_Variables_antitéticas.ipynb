```python
from random import random
import numpy as np
from numba import jit
import matplotlib.pyplot as plt
```

```python
@jit(nopython = True)
def func():
  x1, x2, x = [], [], []
  nr = 10**6
  for i in range(nr):
    r = random()
    a =1/(1+r)
    b =1/(2-r)
    x1.append(a)
    x2.append(b)
    x.append((a+b)/2)
  return x1, x2, x
```

```python
a, b, c = func()
m1 = np.mean(a)
v1 = np.var(a, ddof=1)

m2 = np.mean(b)
v2 = np.var(b, ddof=1)

m = np.mean(c)
v = np.var(c, ddof=1)

print(m1, m2)
print(m)

print(v1, v2)
print(v)

print(50*(1-np.corrcoef(a,b)))
print(100*(v1-v)/v1)
```

```python
@jit(nopython = True)
def gunc():
  x1, x2, x = [], [], []
  nr = 10**6
  for i in range(nr):
    r = random()
    a =1/(1+r**2)
    b =1/(1+(1-r)**2)
    x1.append(a)
    x2.append(b)
    x.append((a+b)/2)
  return x1, x2, x
```

```python
a, b, c = gunc()
m1 = np.mean(a)
v1 = np.var(a, ddof=1)

m2 = np.mean(b)
v2 = np.var(b, ddof=1)

m = np.mean(c)
v = np.var(c, ddof=1)

print(m1, m2)
print(m)

print(v1, v2)
print(v)

print(50*(1-np.corrcoef(a,b)))
print(100*(v1-v)/v1)
```
