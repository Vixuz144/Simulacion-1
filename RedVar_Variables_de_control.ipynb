```python
from random import random
import numpy as np
from scipy.stats import pearsonr
```

```python
def f(x):
  return 1/(1+x)

def g(x):
  return np.sqrt(x)
```

```python
l=[]
li=[]
nr=10**5

for i in range(nr):
  r=random()
  l.append(1/(1+r**2))
  li.append(f(r))
  #li.append(g(r))
print(100*pearsonr(l,li)[0]**2)
```

```python
co=np.cov(l,li)
print(co)

c=-co[0,1]/co[1,1]
print('c =',c)
```

```python
lx, ly, lz = [], [], []

for i in range(nr):
  r=random()
  x = 1/(1+r**2)
  y = f(r)
  #y = g(r)
  z = x + c*(y - np.log(2))
  #z = x + c*(y - 3/2)
  lx.append(x)
  ly.append(y)
  lz.append(z)

mx = np.mean(lx)
my = np.mean(ly)
mz = np.mean(lz)

vx = np.var(lx, ddof=1)
vy = np.var(ly, ddof=1)
vz = np.var(lz, ddof=1)

print(mx,my,mz)
print(vx,vy,vz)
print(100*(vx-vz)/vx)
```
