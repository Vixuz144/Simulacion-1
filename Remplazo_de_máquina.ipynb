```python
import numpy as np
from random import random
import matplotlib.pyplot as plt
from numba import jit
```

```python
def F(alpha):
  return  -np.log(random())/alpha
```

```python
alpha = 1/50
beta = 1/5
nm = 5
s = 3
disp =s
t=0
td = []
for i in range(nm):
  td.append(F(alpha))
tdp = min(td)
tsp = np.inf
l = 0
indicador = 0
while indicador == 0:
  t=min(tdp,tsp)
  if t == tdp:
    if disp==0:
      indicador = 1
      k = td.index(tdp)
    else:
      disp -= 1
      l += 1
      k = td.index(tdp)
      td[k] = t + F(alpha)
      if l == 1:
        tsp = t + F(beta)
      tdp = min(td)
    texto = 'Descomp'
  else:
    l -= 1
    disp += 1
    if l == 0:
      tsp = np.inf
    else:
      tsp = t + F(beta)
    texto = 'Fin Rep'
  print(texto, round(t,5), k, disp, l)
```
