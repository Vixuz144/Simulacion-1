```python
from random import random as rd
import numpy as np
import matplotlib.pyplot as plt
from numba import jit

@jit(nopython = True)
def f(x):
  if 0<= x <1/3:
    return np.sqrt(3*x)
  if 1/3<= x <1:
    return 3 - np.sqrt(6 - 6*x)

@jit(nopython = True)
def ej1(n, f):
  l = list()
  for i in range(n):
    r = rd()
    l.append(f(r))
  return l

if __name__ == "__main__":
  n = 1000000
  l = ej1(n, f)
  plt.hist(l, density = True, bins = 66, color = "cyan", edgecolor = 'k')
  x = list(np.arange(300)/ 100)
  y = []
  for i in range(300):
    y.append(f(x[i]))

  plt.figure()
  plt.plot(x,y)

  print(np.mean(np.log(l)))
```
