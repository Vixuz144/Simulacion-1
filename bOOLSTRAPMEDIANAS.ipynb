```python
from random import random,expovariate,choices
import numpy as np
import matplotlib.pyplot as plt
```

```python
r = 10**5
l = []
for i in range(r):
  l.append(expovariate(3))
m = 10**4
lm = []
for i in range(m):
  l2 = choices(l,k=m)
  lm.append(np.median(l2))

med = np.mean(lm)
desv = np.std(lm, ddof=1)
z= 1.96

print(med)

print(med - z*desv/np.sqrt(m),med + z*desv/np.sqrt(m))
print(np.percentile(lm,5),np.percentile(lm,95))
plt.hist(lm, bins = 20, density = True, color = 'r', edgecolor = 'k')
plt.show()

```

```python
r = 10**3
l = []
for i in range(r):
  # l.append(expovariate(3))
  r= random()
  l.append(r + np.sqrt(r))
m = 10**4
lm = []
for i in range(m):
  l2 = choices(l,k=m)
  # lm.append(np.median(l2))
  lm.append(np.var(l2))

med = np.mean(lm)
var = np.var(lm, ddof=1)
desv = np.std(lm, ddof=1)
z= 1.96

print(med, var)
print()

print(med - z*desv/np.sqrt(m),med + z*desv/np.sqrt(m))
print(np.percentile(lm,5),np.percentile(lm,95))
plt.hist(lm, bins = 20, density = True, color = 'r', edgecolor = 'k')
plt.show()

```
