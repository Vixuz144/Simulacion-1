a) Estime por simulaciòn $P(x>3)$, donde x sigue una distribuciòn normal estandar. Repitalo varias veces y observe si los resultados son constantes

```python
import numpy as np
from random import gauss, random
from numba import jit
```

```python
@jit(nopython = True)
def f(nr= 10**6):
  l = []
  for i in range(nr):
    r = gauss(0,1)
    if r > 3:
      l.append(1)
    else:
      l.append(0)
  return l
```

```python
l = f()
m = np.mean(l)
v = np.var(l, ddof = 1)
print(m, v)
```

b) Use la tècnica de reducciòn de varianza por muestreo importancia para reducir la varianza (use h(x)) ¿En que porcentaje se redujo la varianza?

$h(x) = \left\{\begin{array} ee^{-(x-3)} & \textrm{si } x \geq 3 \\ 0 & \textrm{si } x<3 \end{array}\right.$


```python
@jit(nopython = True)
def h(x):
  if x>3:
    return np.exp(3-x)
  else:
    return 0
```

```python
@jit(nopython = True)
def j(nr=10**6, p=3):
  l,k = [],[]
  for i in range(nr):
    g = gauss(0,1)
    if g > p:
      I = 1
    else:
      I = 0
    l.append(I)

    r = 3 - np.log(random())
    ff = np.exp(-r**2/2)/np.sqrt(2*np.pi)
    if r > p:
      I = 1
    else:
      I = 0
    k.append(I * ff /h(r))
  return l,k
```

```python
r = 10**8
l1, l2 = j(r, 4)
m1 = np.mean(l1)
m2 = np.mean(l2)
v1 = np.var(l1, ddof = 1)
v2 = np.var(l2, ddof = 1)
print(m1, m2)
print(v1, v2)
print(v1/r, v2/r)
print('Reducciòn de varianza:',round(100*(v1-v2)/v1, 2))
```
