```python
from random import random as rd
from statistics import mean, variance

def bino(n,p):
  u = rd()
  pb = (1-p)**n
  suma = pb
  k = 0
  while u > suma:
    pb = ((n-k)/(k+1))*p*pb/(1-p)
    suma = suma + pb
    k += 1
  return k

if __name__ == '__main__':
  n1 = 100000
  m = 20
  p = 1/3
  l = []
  for i in range(n1):
    x = bino(m,p)
    l.append(x)
  print(l.count(0), l.count(20))
  print(mean(l), m*p)
  print(variance(l), m*p*(1-p))
```
